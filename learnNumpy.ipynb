```python
# intersect, difference, union

first = np.array([1,3,7,9,13])
second = np.array([2,3,8,9,14])

ins = np.intersect1d(first,second)
ins

# elements first arr has but not inside second arr
dif = np.setdiff1d(first,second)
dif

# combine together elements unique
uni = np.union1d(first , second)
uni

# loop through first arr elements if it's in the second array
boo = np.isin(first ,second)
boo

```

```python
 # show each array elements and how much time it repeats
arr = np.array([5, 2, 5, 3, 2, 4, 5, 2])

# Get unique elements and their repetition counts
elements, counts = np.unique(arr, return_counts=True)

print("Elements:", elements)
print("Counts:  ", counts)

# Elements: [2 3 4 5]
# Counts:   [3 1 1 3]
```

```python
# import numpy as np

rng = np.random.default_rng()

fruits = np.array(['apple', 'banana', 'cherry'])

# Apple = 70% chance, Banana = 20% chance, Cherry = 10% chance
probabilities = [0.7, 0.2, 0.1]

random_fruits = rng.choice(fruits, size=3, replace=True, p=probabilities)

print(random_fruits)
# Example Output (mostly apples): ['apple', 'banana', 'apple']
```
